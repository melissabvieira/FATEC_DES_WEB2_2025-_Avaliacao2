# Exercício 2 — Estatísticas Linguísticas

Implemente a função abaixo para retornar a quantidade de:

- substantivos;
- verbos;
- adjetivos.


````python
def estatisticas(doc):
    substantivos = 0
    verbos = 0
    adjetivos = 0

    for token in doc:
        if token.pos_ == "NOUN":
            substantivos += 1
        elif token.pos_ == "VERB":
            verbos += 1
        elif token.pos_ == "ADJ":
            adjetivos += 1

    return {
        "substantivos": substantivos,
        "verbos": verbos,
        "adjetivos": adjetivos
    }
    return {}

texto = "The students developed intelligent systems using Python."
doc = nlp(texto)

estatisticas(doc)
````

````python
# Teste
assert isinstance(estatisticas(doc), dict)
print("Estrutura da resposta válida.")
````

# Exercício 3 — Removendo Stop Words

Crie uma nova frase removendo todas as *stop words*.


````python
texto = "Artificial intelligence is changing the way people interact with technology every day."
doc = nlp(texto)

texto_sem_stopwords = []

for token in doc:
    if not token.is_stop:
        texto_sem_stopwords.append(token.text)

print(" ".join(texto_sem_stopwords))
````
